## 2. Run the weekly deck
Runs `run_weekly.bat` (→ `smt_weekly_deck.py` + `looker_capture.py`) and shows progress live.

```python
# Step 1 inside: open Looker once and save all screenshots to output\
# Step 2 inside: build the PowerPoint from those screenshots
import os, subprocess
from pathlib import Path

bat = Path.cwd() / "run_weekly.bat"
if not bat.exists():
    raise FileNotFoundError(f"run_weekly.bat not found in {Path.cwd()}. Open this notebook from the V2 version folder.")

env = dict(os.environ, PYTHONUNBUFFERED="1")  # show messages as they happen
process = subprocess.Popen(
    ["cmd", "/c", str(bat)],
    cwd=bat.parent,
    stdout=subprocess.PIPE,
    stderr=subprocess.STDOUT,
    text=True,
    encoding="utf-8",
    errors="replace",
    env=env,
)
for line in process.stdout:
    print(line, end="", flush=True)
exit_code = process.wait()
if exit_code != 0:
    raise RuntimeError(f"run_weekly.bat stopped with an error (exit code {exit_code}). See the messages above.")

decks = sorted((Path.cwd() / "output").rglob("*.pptx"), key=lambda deck: deck.stat().st_mtime)
print("\nDone. Newest deck:", decks[-1] if decks else "none found")
```
